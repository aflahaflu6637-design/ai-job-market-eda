# Bivariate Analysis

## Objective

The objective of this notebook is to explore relationships between variables in the AI & Data Science Job Market dataset.

The analysis includes:

- Salary vs Experience Level
- Salary vs Job Title
- Salary vs Country
- Salary vs Industry
- Salary vs Remote Type

The goal is to generate business insights and identify factors influencing salary trends.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

print("Libraries imported successfully")
```

```python
df = pd.read_csv("../data/interim/cleaned_day2.csv")

print("Dataset loaded successfully")
```

```python
df["average_salary"] = (
    df["salary_min_usd"] +
    df["salary_max_usd"]
) / 2

df.head()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="experience_level",
    y="average_salary",
    data=df
)

plt.title("Average Salary by Experience Level")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x="job_title",
    y="average_salary",
    data=df
)

plt.xticks(rotation=45)

plt.title("Average Salary by Job Title")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x="country",
    y="average_salary",
    data=df
)

plt.title("Average Salary by Country")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x="industry",
    y="average_salary",
    data=df
)

plt.title("Average Salary by Industry")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="remote_type",
    y="average_salary",
    data=df
)

plt.title("Average Salary by Work Type")

plt.show()
```

```python
numerical_df = df.select_dtypes(include=["int64", "float64"])

numerical_df.corr()
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    numerical_df.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.show()
```
