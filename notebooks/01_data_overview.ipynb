# AI & Data Science Job Market Analysis (2020–2026)

## Business Understanding

The Artificial Intelligence and Data Science industry has experienced rapid growth in recent years. Organizations across different industries are actively hiring AI professionals with varying salary packages, experience requirements, and work arrangements.

This project aims to perform Exploratory Data Analysis (EDA) on an AI & Data Science Job Market dataset to understand hiring patterns, salary trends, company characteristics, and workforce distribution.

### Objectives

- Analyze salary trends across job roles and experience levels
- Study remote work and employment patterns
- Examine industry-wise and country-wise hiring trends
- Identify factors affecting salary ranges
- Detect data quality issues
- Generate meaningful insights from the AI job market
