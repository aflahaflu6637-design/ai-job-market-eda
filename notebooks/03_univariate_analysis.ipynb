# Univariate Analysis

## Objective

The objective of this notebook is to analyze individual variables in the AI & Data Science Job Market dataset.

The analysis includes:

- Numerical variable distributions
- Categorical variable distributions
- Frequency analysis
- Summary statistics
- Initial business insights

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

print("Libraries imported successfully")
```

```python
df = pd.read_csv("../data/interim/cleaned_day2.csv")

print("Dataset loaded successfully")
```

```python
df[
    [
        "min_experience_years",
        "salary_min_usd",
        "salary_max_usd",
        "posted_year"
    ]
].describe()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["salary_min_usd"], bins=30, kde=True)

plt.title("Distribution of Minimum Salary")
plt.xlabel("Minimum Salary (USD)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["salary_max_usd"], bins=30, kde=True)

plt.title("Distribution of Maximum Salary")
plt.xlabel("Maximum Salary (USD)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="experience_level", data=df)

plt.title("Experience Level Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x="job_title", data=df)

plt.title("Job Title Distribution")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="remote_type", data=df)

plt.title("Remote Work Distribution")

plt.show()
```

# Numerical Variable Analysis

## Minimum Salary Distribution

### Observation

- The minimum salary ranges from approximately $50,000 to $155,000.
- The distribution shows multiple salary clusters.
- No extreme salary outliers are visible.
- Salary opportunities appear distributed across low, medium, and high salary ranges.

## Maximum Salary Distribution

### Observation

- Maximum salaries range from approximately $65,000 to $180,000.
- Multiple salary bands are observed.
- The distribution suggests varying compensation levels across AI-related roles.
- No abnormal salary spikes are detected.
